# 02 - Model Experiments & Recall Comparison
This notebook demonstrates training and comparing Logistic Regression, Random Forest, and XGBoost with primary emphasis on **Recall**.

```python
import sys
import json
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

with open(PROJECT_ROOT / "models" / "evaluation_metrics.json") as f:
    metrics = json.load(f)

print(f"Primary Metric: {metrics['primary_metric']}")
print(f"Champion Model: {metrics['best_model']}")
for m_name, m_stats in metrics["models"].items():
    print(f"- {m_name}: Recall={m_stats['recall']:.2%}, F1={m_stats['f1_score']:.2%}, AUC={m_stats['roc_auc']:.3f}")
```
